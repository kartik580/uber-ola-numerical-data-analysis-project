```python
user_id = np.arange(1, 26)
rider_id = np.arange(101, 126)
experience=np.random.randint(1,18,25)
rating= np.random.randint(1,6,25)
total_trip_time=np.random.randint(0,745,25)
no_of_trip= np.random.randint(15,451,25)
distance_travelled=np.random.randint(15,12000,25)
vehicle_type=np.random.randint(2,5,25)

uber_data = np.concatenate((user_id.reshape(-1,1),
                        rider_id.reshape(-1,1),
                        experience.reshape(-1,1),
                        rating.reshape(-1,1),
                        total_trip_time.reshape(-1,1),
                        no_of_trip.reshape(-1,1),
                        distance_travelled.reshape(-1,1),
                        vehicle_type.reshape(-1,1)),axis=1)

print(uber_data)

```

```python
data=np.column_stack([user_id,
                      rider_id,
                      experience,
                      rating,
                      total_trip_time,
                      no_of_trip,
                      distance_travelled,
                      vehicle_type])
print(data)
```
